```python
from invoke_agent.core import execute_api_call
import base64

raw_email = """To: cooperdoyle@gmail.com
Subject: Hello from API
Content-Type: text/plain; charset="UTF-8"
Content-Transfer-Encoding: 7bit
MIME-Version: 1.0

This is the body of the email."""

# base64url encode (remove padding '=')
encoded = base64.urlsafe_b64encode(raw_email.encode("utf-8")).decode("utf-8").rstrip("=")
print(encoded)

execute_api_call({'action': 'POST',
 'url': 'https://www.googleapis.com/gmail/v1/users/me/messages/send',
 'auth_code': 'oauth::Bearer::i',
 'parameters': {'raw': encoded}
})
```
